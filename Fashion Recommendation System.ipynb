```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from google.colab import files
```

```python
print("Upload fashion_products.csv")

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

df.head()
```

```python
print("Dataset Shape:", df.shape)

df.head(10)
```

```python
df = df.fillna("")

df["combined_features"] = (
    df["Category"].astype(str)
    + " "
    + df["Color"].astype(str)
    + " "
    + df["Style"].astype(str)
    + " "
    + df["Brand"].astype(str)
)

df[["Product_Name","combined_features"]].head()
```

```python
vectorizer = TfidfVectorizer()

feature_vectors = vectorizer.fit_transform(
    df["combined_features"]
)

print("Feature Extraction Completed")
```

```python
similarity = cosine_similarity(
    feature_vectors
)

print("Similarity Matrix Created")
```

```python
def recommend(product_name):

    matched = df[
        df["Product_Name"]
        .str.contains(
            product_name,
            case=False,
            na=False
        )
    ]

    if len(matched) == 0:
        print("Product Not Found")
        return

    index = matched.index[0]

    similarity_scores = list(
        enumerate(similarity[index])
    )

    sorted_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    print("\nTop Recommended Products:\n")

    count = 0

    for i, score in sorted_scores[1:]:

        print(
            "Product:",
            df.iloc[i]["Product_Name"]
        )

        print(
            "Category:",
            df.iloc[i]["Category"]
        )

        print(
            "Brand:",
            df.iloc[i]["Brand"]
        )

        print(
            "Price:",
            "₹",
            df.iloc[i]["Price_INR"]
        )

        print(
            "Rating:",
            df.iloc[i]["Rating"]
        )

        print("---------------------")

        count += 1

        if count == 5:
            break
```

```python
product = input(
    "Enter Product Name: "
)

recommend(product)
```
